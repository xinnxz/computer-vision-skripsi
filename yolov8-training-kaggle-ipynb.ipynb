## 2. Mulai Training (Pelatihan AI)
Pastikan di dalam folder `dataset` terdapat file `data.yaml`. File ini berisi informasi kelas bahan baku.
Kita akan melatih dengan `epochs=50`. Semakin tinggi epoch, semakin pintar AI-nya, tapi waktu latihnya lebih lama.

```python
# 2. Panggil YOLO
from ultralytics import YOLO
# 3. Gunakan model EXTRA LARGE (Paling Kuat)
model = YOLO('yolov8x.pt')
# 4. Mulai Training (Perhatikan path data.yaml nya sudah saya sesuaikan dengan foto Mas Luthfi)
results = model.train(
    data='/kaggle/input/datasets/luthfialfa/dataset-gudang/data.yaml',  
    epochs=300,          # Belajar 300 putaran
    patience=50,         
    imgsz=640,           
    batch=8,             # Batch diturunkan ke 8 biar Kaggle tidak error
    name='model_gudang_lisna_X',
    optimizer='auto'
)
```

## 3. Validasi Hasil (Opsional)
Lihat seberapa bagus hasil deteksinya pada data validasi.

```python
metrics = model.val()
print(metrics)
```
